### Finding The Frequencies of notes in different chords and scales.

Background:

Musical chords and scales are defined by the relationships between different notes and the patterns in between them. 

The type of tuning that I am using is called just intonation and is based on the ratios between notes.

How it works:
    
Major chords are defined in just intonation as having the ratio 1:5/4:3/2

Minor chordes are defined in just intonation as having the ratio 1:6/5:3/2

Diminished chords are defined in just intonation as having the ratio 1:6/5:45/32



We find our initial notes using the ratios 

```python
a = 440
bb = a*(2)**(1/12)
b = bb*(2)**(1/12)
c = b*(2)**(1/12)
db = c*(2)**(1/12)
d = db*(2)**(1/12)
eb= d*(2)**(1/12)
e=eb*(2)**(1/12)
f=e*(2)**(1/12)
gb=f*(2)**(1/12)
g=gb*(2)**(1/12)
ab=g*(2)**(1/12)


print("A = ",a, "hertz")
print("Bb =",bb, "hertz")
print("B =", b, "hertz")
print("...")

```

We can then print in the notes in a major chord, for example C

```python
Root = c
Third = Root * 5/4
Fifth = Root * 3/2

print("C Major Chord")
print(Root, "hertz")
print(Third, "hertz")
print(Fifth, "hertz")
```

We can also print a minor chord, for example A

```python
root = a
third = root * 5/4
fifth = root * 3/2

print("A Minor Chord")
print(root, "hertz")
print(third, "hertz")
print(fifth, "hertz")
```
